```python
from dataset import get_dataloader
from torchvision.transforms import ToPILImage
```

```python
root_dir = "./data/faces_webface_112x112"
local_rank = 0
batch_size = 32
dali = False
dali_aug = False
seed = 2048
num_workers = 2
```

```python
train_loader = get_dataloader(
    root_dir=root_dir,
    local_rank=local_rank,
    batch_size=batch_size,
    dali=dali,
    dali_aug=dali_aug,
    seed=seed,
    num_workers=num_workers
)
```

```python
def tensor_to_pil(sample):
    sample = (sample * 0.5) + 0.5
    return ToPILImage()(sample.clamp(0, 1))
```
