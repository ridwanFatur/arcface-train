```python
from pathlib import Path
import zipfile
import gdown
```

```python
URL = "https://drive.google.com/file/d/1KxNCrXzln0lal3N4JiYl9cFOIhT78y1l/view"
```

```python
def download(url: str, output_path: str = "data/dataset.zip") -> Path:
    output_path = Path(output_path)
    output_path.parent.mkdir(parents=True, exist_ok=True)

    # Download the file from Google Drive.
    downloaded_path = gdown.download(url, str(output_path), quiet=False)

    if downloaded_path is None:
        raise RuntimeError("Failed to download the file from Google Drive.")

    print(f"Download completed: {output_path.resolve()}")
    return output_path


def extract(zip_path: str, output_dir: str = "data") -> None:
    zip_path = Path(zip_path)
    output_path = Path(output_dir)
    output_path.mkdir(parents=True, exist_ok=True)

    # Validate the ZIP archive.
    if not zipfile.is_zipfile(zip_path):
        raise ValueError("The downloaded file is not a valid ZIP archive.")

    # Extract the ZIP archive into the output directory.
    with zipfile.ZipFile(zip_path, "r") as zip_ref:
        zip_ref.extractall(output_path)

    print(f"Extraction completed: {output_path.resolve()}")
```
