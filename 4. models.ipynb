```python
import sys
from pathlib import Path

ROOT = Path("./arcface_torch")
sys.path.insert(0, str(ROOT))
```

```python
import pandas as pd
from backbones import get_model
```

```python
model_names = [
    'r18',
    'r34',
    'r50',
    'r100',
    'r200',
    'r2060',
    'mbf',
    'mbf_large',
    'vit_t',
    'vit_t_dp005_mask0',
    'vit_s',
    'vit_s_dp005_mask_0',
    'vit_b',
    'vit_b_dp005_mask_005',
    'vit_l_dp005_mask_005',
    'vit_h'
]
```

```python
results = []
for model_name in model_names:
    backbone = get_model(
        model_name,
        dropout=0.0,
        fp16=False,
        num_features=512
    )
    num_params = sum(p.numel() for p in backbone.parameters())
    results.append({
        "model": model_name,
        "parameters": num_params,
        "parameters_M": num_params / 1e6
    })
```

```python
df = pd.DataFrame(results).sort_values("parameters", ascending=False).reset_index(drop=True)
df["parameters"] = df["parameters"].map("{:,}".format)
```
